```python
#Escolha o campo da nota que se quer adicionar
campo_nota = 1

#Escolha a coluna da tabela que se quer adicionar
coluna_nota = 'NOTA'

local = os.getcwd()

Tk().withdraw() 
nome_arquivo = askopenfilename(initialdir = local)
turmas = pd.ExcelFile(nome_arquivo).sheet_names
turmas


campo_nota = (campo_nota-1)*4

for turma in turmas:   
    df = pd.read_excel(nome_arquivo, sheet_name=turma)
    ids_turma = pd.read_excel('caminho_lancar_notas.xlsx')
    link_lancar = ids_turma[ids_turma['turma'] == turma].links
    link_lancar = str(link_lancar.values[0])
    nami.get(link_lancar)
    wdw = WebDriverWait(nami, 10)


    try:        
        botao = WebDriverWait(nami, 10).until(EC.element_to_be_clickable((By.CSS_SELECTOR, 'a.btn.btn-warning')))
        botao.click()
        sleep(2)
    except:        
        print('botão "abrir apuração de notas" não está visível ou clicável')
        sleep(2)
  
   
  
    t = nami.find_elements(By.CSS_SELECTOR, 'tbody')        
    l = t[1].find_elements(By.CSS_SELECTOR, 'tr')
    
    alunos_totais = np.arange(len(l))
    wdw = WebDriverWait(nami, 10)
    print(turma)

    
    for idx,  nota, nome in zip(df.IDX, df[coluna_nota], df['NOME DO ALUNO']):     
        try: 
            insere_nota(campo_nota, idx, nota, l)
        except:
            print(nome)
            pass

    nami.find_element(By.CSS_SELECTOR, 'div[class="bg-success legend"]').click()
    sleep(2)

    try:        
        nami.find_element(By.CSS_SELECTOR, 'a[class="btn btn-primary"]').click()
        sleep(0.1)
    except:        
        print(f'houve um problema no botão de finalizar notas da turma {turma}')
        sleep(0.1)
        pass
```
